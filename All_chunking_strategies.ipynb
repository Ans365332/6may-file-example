```python
# Chunking
```

```python
sample_text = """
Artificial Intelligence (AI) is transforming industries by automating repetitive tasks and improving decision-making.

Machine Learning is a subset of Artificial Intelligence that enables systems to learn patterns from data without being explicitly programmed.

Deep Learning is a specialized branch of Machine Learning that uses neural networks with multiple layers.

Natural Language Processing (NLP) enables computers to understand, interpret, and generate human language.

Python is one of the most popular programming languages for Artificial Intelligence, Machine Learning, and Data Science because of its rich ecosystem of libraries.

Retrieval-Augmented Generation (RAG) combines Large Language Models with external knowledge sources to generate accurate and context-aware responses.
"""
```

```python
# 1. Character Chunking
```

```python
print("="*60)
print("Character Chunking")
print("="*60)

chunk_size = 100

chunks = [
    sample_text[i:i+chunk_size]
    for i in range(0, len(sample_text), chunk_size)
]

for i, chunk in enumerate(chunks, 1):
    print(f"\nChunk {i}")
    print(chunk)
```

```python
# 2. Word Chunking
```

```python
print("="*60)
print("Word Chunking")
print("="*60)

chunk_size = 20

words = sample_text.split()

chunks = [
    " ".join(words[i:i+chunk_size])
    for i in range(0, len(words), chunk_size)
]

for i, chunk in enumerate(chunks, 1):
    print(f"\nChunk {i}")
    print(chunk)
```

```python
# 5. Fixed Size Chunking
```

```python
print("="*60)
print("Fixed Size Chunking")
print("="*60)

chunk_size = 150

chunks = []

start = 0

while start < len(sample_text):

    end = start + chunk_size

    chunks.append(sample_text[start:end])

    start = end

for i, chunk in enumerate(chunks,1):

    print(f"\nChunk {i}")

    print(chunk)
```

```python
# 6. Sliding Window (Overlap Chunking)
```

```python
print("="*60)
print("Sliding Window Chunking")
print("="*60)

chunk_size = 150

overlap = 40

chunks = []

start = 0

while start < len(sample_text):

    end = start + chunk_size

    chunks.append(sample_text[start:end])

    start += (chunk_size - overlap)

for i, chunk in enumerate(chunks,1):

    print(f"\nChunk {i}")

    print(chunk)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
Artificial Intelligence is transforming industries.

Machine Learning is a subset of AI.

Deep Learning uses Neural Networks.

Python is the most popular language for AI.

RAG combines LLMs with external knowledge.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=80,
    chunk_overlap=20
)

chunks = splitter.split_text(text)

print("Total Chunks :", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}")
    print(chunk)
```

```python
# 8. Token Chunking (tiktoken)
```

```python
!pip install -q tiktoken
```

```python
import tiktoken

text = """
Artificial Intelligence is transforming industries.
Machine Learning is a subset of AI.
Deep Learning uses Neural Networks.
"""

enc = tiktoken.get_encoding("cl100k_base")

tokens = enc.encode(text)

chunk_size = 20

chunks = [
    tokens[i:i+chunk_size]
    for i in range(0, len(tokens), chunk_size)
]

print("Total Token Chunks :", len(chunks))

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}")

    print(enc.decode(chunk))
```

```python
# 9.  Semantic Chunking
```

```python
!pip install -q sentence-transformers scikit-learn nltk
```

```python
import nltk
nltk.download("punkt")
```

```python
from nltk.tokenize import sent_tokenize

from sentence_transformers import SentenceTransformer

from sklearn.metrics.pairwise import cosine_similarity

text = """
Artificial Intelligence is transforming industries.

Machine Learning is a subset of AI.

Deep Learning uses neural networks.

Football is a popular sport.

Messi is one of the greatest football players.
"""

sentences = sent_tokenize(text)

model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(sentences)

print("="*50)

for i in range(len(sentences)-1):

    sim = cosine_similarity(
        [embeddings[i]],
        [embeddings[i+1]]
    )[0][0]

    print(f"\nSentence {i+1}")

    print(sentences[i])

    print("\nSimilarity :", round(sim,3))
```

```python
# 10. Markdown Header Chunking
```

```python
!pip install -q langchain-text-splitters
```

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

markdown = """

# AI

Artificial Intelligence Introduction

## Machine Learning

Machine Learning Notes

## Deep Learning

Deep Learning Notes

# Python

Python Basics

"""

headers = [

("#","H1"),

("##","H2")

]

splitter = MarkdownHeaderTextSplitter(headers)

chunks = splitter.split_text(markdown)

for chunk in chunks:

    print("="*50)

    print(chunk.page_content)

    print(chunk.metadata)
```

```python
# 11. HTML Header Chunking
```

```python
from bs4 import BeautifulSoup

html = """

<h1>Artificial Intelligence</h1>

<p>AI Introduction</p>

<h2>Machine Learning</h2>

<p>ML Notes</p>

<h2>Deep Learning</h2>

<p>DL Notes</p>

"""

soup = BeautifulSoup(html,"html.parser")

for tag in soup.find_all(["h1","h2","p"]):

    print(tag.name,"=>",tag.text)
```

```python
# 12. Code Chunking
```

```python
import ast

code = """

def train():

    print("Training")

def predict():

    print("Prediction")

class Student:

    pass

"""

tree = ast.parse(code)

for node in tree.body:

    if isinstance(node, ast.FunctionDef):

        print("Function :", node.name)

    elif isinstance(node, ast.ClassDef):

        print("Class :", node.name)
```

```python
# 14. Hierarchical Chunking
```

```python
document = {

"Chapter 1":{

"Section 1":[

"AI",

"History"

],

"Section 2":[

"Machine Learning"

]

},

"Chapter 2":{

"Section 1":[

"Deep Learning"

]

}

}

for chapter,sections in document.items():

    print(chapter)

    for section,content in sections.items():

        print("  ",section)

        print("     ",content)
```

```python
# 15. Parent Child Chunking
```

```python
parent = """

Artificial Intelligence

Machine Learning

Deep Learning

Python

RAG

"""

children = parent.split("\n")

children = [c for c in children if c]

print("Parent Chunk\n")

print(parent)

print("\nChild Chunks\n")

for child in children:

    print(child)
```
