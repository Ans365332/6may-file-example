```python
# Hybrid Embedding
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "Apple stock market",
    "Apple fruit benefits",
    "Tesla electric cars"
]

query = "Apple stock"

# --------------------------
# Dense Embedding
# --------------------------

model = SentenceTransformer("all-MiniLM-L6-v2")

dense_vectors = model.encode(documents)

query_dense = model.encode(query)
print("Dense Embedding Shape")
print(dense_vectors.shape)


# --------------------------
# Sparse Embedding (TF-IDF)
# --------------------------

tfidf = TfidfVectorizer()

sparse_vectors = tfidf.fit_transform(documents)

query_sparse = tfidf.transform([query])

print("\nSparse Matrix Shape")
print(sparse_vectors.shape)
```

```python
# Semantic Search using Embeddings
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

documents = [
    "Dogs are loyal animals.",
    "Cats love drinking milk.",
    "Artificial Intelligence is transforming industries.",
    "Python is a programming language.",
    "Machine Learning is a subset of AI."
]

query = "What is Artificial Intelligence?"

# Generate embeddings
doc_embeddings = model.encode(documents)
query_embedding = model.encode([query])

# Calculate similarities
scores = cosine_similarity(query_embedding, doc_embeddings)[0]

print("Similarity Scores\n")

for doc, score in zip(documents, scores):
    print(f"{score:.4f} ---> {doc}")

best_index = np.argmax(scores)

print("\nMost Relevant Document\n")
print(documents[best_index])
```

```python
# Task : Top-3 Similar Documents (Top-K Retrieval)

# Task Description:

# Build a simple semantic search application using Python and the Sentence Transformers library. Create a dataset containing 30 text documents
# on different topics. When a user enters a search query, generate embeddings for both the documents and the query, calculate the Cosine Similarity
# between the query and each document, rank the documents based on their similarity scores, and display the Top 3 most relevant documents along with
# their similarity scores. Do not use any LLM, Vector Database, or LangChain. The goal of this task is to understand how the retrieval stage of a
#  RAG system works by implementing semantic search from scratch using embeddings and similarity search.
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Dataset (30 Documents)
documents = [
    "Dogs are loyal animals.",
    "Cats love drinking milk.",
    "Artificial Intelligence is transforming industries.",
    "Python is a programming language.",
    "Machine Learning is a subset of AI.",
    "Deep Learning uses neural networks.",
    "Data Science combines statistics and programming.",
    "SQL is used to manage relational databases.",
    "MongoDB is a NoSQL database.",
    "FastAPI is used for building APIs.",
    "Flask is a lightweight web framework.",
    "Git is a version control system.",
    "Docker is used for containerization.",
    "Kubernetes manages containers.",
    "Cloud Computing provides online resources.",
    "Cybersecurity protects computer systems.",
    "Natural Language Processing understands text.",
    "Computer Vision analyzes images.",
    "Pandas is used for data analysis.",
    "NumPy is used for numerical computing.",
    "Matplotlib creates graphs and charts.",
    "Cricket is a popular sport in India.",
    "Football is played worldwide.",
    "The Taj Mahal is in Agra.",
    "The Earth revolves around the Sun.",
    "Renewable energy includes solar power.",
    "Electric vehicles reduce pollution.",
    "Exercise improves physical health.",
    "Drinking water keeps the body hydrated.",
    "Pizza is a famous Italian food."
]

# User Query
query = input("Enter your search query: ")

# Generate Embeddings
doc_embeddings = model.encode(documents)
query_embedding = model.encode([query])

# Calculate Cosine Similarity
scores = cosine_similarity(query_embedding, doc_embeddings)[0]

print("\nSimilarity Scores\n")

for doc, score in zip(documents, scores):
    print(f"{score:.4f} ---> {doc}")

# Top 3 Similar Documents
top_k = 3
top_indices = np.argsort(scores)[::-1][:top_k]

print("\nTop 3 Similar Documents\n")

for rank, index in enumerate(top_indices, start=1):
    print(f"Rank {rank}")
    print(f"Similarity Score : {scores[index]:.4f}")
    print(f"Document         : {documents[index]}")
    print("-" * 60)
```
